```python
#센서 컬럼만 선택
sensor_cols = [str(i) for i in range(590)]

# 모든 행의 값이 실제 숫자 0 인 컬럼
all_zero_columns = [
    col for col in sensor_cols
    if df[col].notna().all() and (df[col] == 0).all()
]

print("All-Zero Columns:", all_zero_columns)
print("Number of All-Zero Columns:", len(all_zero_columns))
```

```python
# NaN을 제외했을 때 관측값 종류가 1개인 센서 컬럼 찾기
observed_constant_columns = [
    col for col in sensor_cols
    if df[col].dropna().nunique() == 1
]

print("Observed-value Constant Columns:", observed_constant_columns)
print("Number of Observed-value Constant Columns:", len(observed_constant_columns))
```

```python
# 관측값 1개인게 대체 무슨 값이고, 각 컬럼에는 NaN이 얼마나 섞여 있는가?

constant_summary = []

for col in observed_constant_columns:
    observed_value = df[col].dropna().iloc[0]
    missing_count = df[col].isna().sum()
    missing_rate = df[col].isna().mean() * 100

    constant_summary.append({
        "Feature": col,
        "Observed_Value": observed_value,
        "Missing_Count": missing_count,
        "Missing_Rate(%)": missing_rate
    })

constant_summary_df = pd.DataFrame(constant_summary)

display(constant_summary_df.head(20))
```

```python
# NaN까지 포함해서 실제로 모든 행이 동일한 Strict Constant Feature
strict_constant_columns = [
    col for col in sensor_cols
    if df[col].notna().all() and df[col].nunique() == 1
]

# 관측값은 하나지만 일부 NaN이 존재하는 Feature
constant_with_missing_columns = [
    col for col in observed_constant_columns
    if df[col].isna().any()
]

print("Strict Constant Columns:", strict_constant_columns)
print("Number of Strict Constant Columns:", len(strict_constant_columns))

print()

print("Observed Constant + Missing Columns:", constant_with_missing_columns)
print("Number of Observed Constant + Missing Columns:", len(constant_with_missing_columns))
```

```python
# RAW Sensor Data에서 실제 숫자 0확인

sensor_df = df[sensor_cols]

# 전체 0 개수
total_zero_count = (sensor_df == 0).sum().sum()

# 0이 한 번이라도 증장하는 Feature 개수
features_with_zero = (sensor_df == 0).any().sum()

print("Total Zero Values:", total_zero_count)
print("Features Containing Zero:", features_with_zero)
```

## Phase 1. Raw Data Audit Summary

-Dataset: 1,567 samples * 592 columns
-Process features: 590 anonymous measurement variables(secom 데이터의 본래 가진 특성. 변수의 익명화)
-Pass (-1): 1,463 samples (93.36%)
-Fail (1): 104 samples (6.64%)
-Overall missing values: 41,951 (4.52%) 결측치
-Maximum feature-level missing rate: 91.19% (한 열의 91.19%가 결측치임. 모든 행이 NaN인 열은 존재하지 않음)
-All-missing features: 0
-Observed-value constant features with missing values: 116(결측치를 제외한 나머지 행이 모두 같은 값인 열의 개수)
-Raw zero values in process features: 200,005 (21.6%) 전체 데이터셋에서, 값이 0인 셀의 개수
-Features containing at least one zero: 215
-Missing values in Time: 0 (시간 열에는 문제없다)
-Missing values in Pass/ Fail: 0 
-Target values: -1, 1

### Key Findings

1. The dataset shows a strong class imbalance, with Fail samples accounting for approximately 6.64%.
2. Although the overall missing rate is 4.52%, missing values are highly concentrated in some features, with the highest missing rate exceeding 91%.
3. No all-missing or strict constant features were identified.
4. 116 features have only one observed numerical value but also contain missing values. These features will not be automatically removed because their missingness patterns may contain predictive information.
5. Raw process data already contains a substantial number of numerical zero values. Therefore, missing values must be distinguished from actual zero values during preprocessing.
