```python
import pandas as pd

# Raw Dara 불러오기
df = pd.read_csv("../01_raw/uci-secom.csv")

# 익명화 공정 측정 변수
sensor_cols = [str(i) for i in range(590)]

sensor_df = df[sensor_cols]
```

```python
# Feature별 Missing 정보 요약

sensor_missing_summary = pd.DataFrame({
    "Feature_ID": sensor_cols,
    "Missing_Count": sensor_df.isna().sum().values,
    "Missing_Rate_pct": (sensor_df.isna().mean() * 100).values
})

# Missing Rate가 높은 순서대로 정렬
sensor_missing_summary = sensor_missing_summary.sort_values(
    "Missing_Rate_pct",
    ascending=False
)

display(sensor_missing_summary.head(20))
```

```python
print("Number of Features:", len(sensor_missing_summary))
print("Total Sensor Missing:", sensor_missing_summary["Missing_Count"].sum())
```

```python
sensor_missing_summary.to_csv(
    "../03_output/sensor_missing_summary.csv",
    index=False
)
```
